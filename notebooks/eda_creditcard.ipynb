```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
df["Amount"].hist(bins=50, ax=axes[0])
axes[0].set_title("Amount (raw)")
import numpy as np
np.log1p(df["Amount"]).hist(bins=50, ax=axes[1])
axes[1].set_title("Amount (log1p)")
plt.tight_layout()
plt.show()
```
